```python
df5= df_new[df_new["script"] == "LlamaIndex_Hybrid_Rerank"]
import pandas as pd




category_rank = {"TP": 2, "FP": 1}

df5["correctness_rank"] = df5["correctness_category"].map(category_rank).fillna(0).astype(int)


df5["faithfulness_1to5"] = pd.to_numeric(df5["faithfulness_1to5"], errors="coerce").fillna(-1)
df5["answer_relevance_1to5"] = pd.to_numeric(df5["answer_relevance_1to5"], errors="coerce").fillna(-1)
df5["helpfulness_final_1to5"] = pd.to_numeric(df5["helpfulness_final_1to5"], errors="coerce").fillna(-1)


df_sorted5 = df5.sort_values(
    by=[
        "script",
        "question_id",
        "correctness_rank",      # TP > FP > Rest
        "faithfulness_1to5",     # höher besser
        "answer_relevance_1to5", # höher besser
        "helpfulness_final_1to5" # optional tie-breaker
    ],
    ascending=[True, True, False, False, False, False],
)


df_best5 = df_sorted5.drop_duplicates(subset=["script", "question_id"], keep="first")
df5= df_new[df_new["script"] == "LlamaIndex_Hybrid_Rerank"]

```

```python
df3= df_new[df_new["script"] == "SimpleKG_Hybrid_Rerank"]
import pandas as pd




category_rank = {"TP": 2, "FP": 1}

df3["correctness_rank"] = df3["correctness_category"].map(category_rank).fillna(0).astype(int)


df3["faithfulness_1to5"] = pd.to_numeric(df3["faithfulness_1to5"], errors="coerce").fillna(-1)
df3["answer_relevance_1to5"] = pd.to_numeric(df3["answer_relevance_1to5"], errors="coerce").fillna(-1)

df3["helpfulness_final_1to5"] = pd.to_numeric(df3["helpfulness_final_1to5"], errors="coerce").fillna(-1)

df_sorted3 = df3.sort_values(
    by=[
        "script",
        "question_id",
        "correctness_rank",      # TP > FP > Rest
        "faithfulness_1to5",     # höher besser
        "answer_relevance_1to5", # höher besser
        "helpfulness_final_1to5" # optional tie-breaker
    ],
    ascending=[True, True, False, False, False, False],
)


df_best3 = df_sorted3.drop_duplicates(subset=["script", "question_id"], keep="first")
```

```python
df2= df_new[df_new["script"] == "SimpleKG_Hybrid"]
import pandas as pd



category_rank = {"TP": 2, "FP": 1}

df2["correctness_rank"] = df2["correctness_category"].map(category_rank).fillna(0).astype(int)


df2["faithfulness_1to5"] = pd.to_numeric(df2["faithfulness_1to5"], errors="coerce").fillna(-1)
df2["answer_relevance_1to5"] = pd.to_numeric(df2["answer_relevance_1to5"], errors="coerce").fillna(-1)

df2["helpfulness_final_1to5"] = pd.to_numeric(df2["helpfulness_final_1to5"], errors="coerce").fillna(-1)


df_sorted2 = df2.sort_values(
    by=[
        "script",
        "question_id",
        "correctness_rank",      # TP > FP > Rest
        "faithfulness_1to5",     # höher besser
        "answer_relevance_1to5", # höher besser
        "helpfulness_final_1to5" # optional tie-breaker
    ],
    ascending=[True, True, False, False, False, False],
)


df_best2 = df_sorted2.drop_duplicates(subset=["script", "question_id"], keep="first")
```

```python
df= df_new[df_new["script"] == "LLMGraph_Hybrid"]
import pandas as pd



# 1) Priorität für correctness_category festlegen (höher = besser)
# TP soll gewinnen, danach FP, danach alles andere
category_rank = {"TP": 2, "FP": 1}

df["correctness_rank"] = df["correctness_category"].map(category_rank).fillna(0).astype(int)

# 2) Scores robust in Zahlen umwandeln (falls sie als Strings vorliegen)
df["faithfulness_1to5"] = pd.to_numeric(df["faithfulness_1to5"], errors="coerce").fillna(-1)
df["answer_relevance_1to5"] = pd.to_numeric(df["answer_relevance_1to5"], errors="coerce").fillna(-1)

# (Optional) weiterer Tie-Breaker, falls du willst:
df["helpfulness_final_1to5"] = pd.to_numeric(df["helpfulness_final_1to5"], errors="coerce").fillna(-1)

# 3) Sortieren: pro (script, question_id) die beste Zeile nach Ranking zuerst
df_sorted = df.sort_values(
    by=[
        "script",
        "question_id",
        "correctness_rank",      # TP > FP > Rest
        "faithfulness_1to5",     # höher besser
        "answer_relevance_1to5", # höher besser
        "helpfulness_final_1to5" # optional tie-breaker
    ],
    ascending=[True, True, False, False, False, False],
)

# 4) Pro (script, question_id) nur die erste (beste) Zeile behalten
df_best = df_sorted.drop_duplicates(subset=["script", "question_id"], keep="first")


```

```python
df_newest = df_new[df_new["script"] != "LLMGraph_Hybrid"]
df_newest2 = df_newest[df_newest["script"] != "SimpleKG_Hybrid"] 
df_newest3 = df_newest2[df_newest2["script"] != "SimpleKG_Hybrid_Rerank"] 
df_newest4 = df_newest3[df_newest3["script"] != "LlamaIndex_Dense"] 
df_newest5 = df_newest4[df_newest4["script"] != "LlamaIndex_Hybrid_Rerank"] 


df_newest_combi = (
    pd.concat(
        [df_newest5,df_best5, df_best4,df_best3, df_best2, df_best ],
        axis=0
    )
    .drop_duplicates()
)
df_newest_combi=  df_newest_combi[df_newest_combi["query_type"] != "relational"]
df_newest_combi["script"] = df_newest_combi["script"].replace({
    "LlamaIndex_Dense_Community_Rerank": "LlamaIndex_Hybrid_Community"
})
df_newest_combi["script"] = df_newest_combi["script"].replace({
    "SimpleKG_Hybrid_Community_Rerank": "SimpleKG_Hybrid_Community"
})
df_newest_combi["script"] = df_newest_combi["script"].replace({
    "Community_Vector_Retriever_HyDE": "LlamaIndex_Community"
})

```

```python
df_newest_combi = df_newest_combi.drop_duplicates(
    subset=["script", "question_id"],
    keep="first"
)
df_newest_combi = df_newest_combi[
    (df_newest_combi["script"] != "SimpleKG_Vector_KG_Retriever") &
    (df_newest_combi["script"] != "LlamaIndex_Hybrid_Community")
]
```
