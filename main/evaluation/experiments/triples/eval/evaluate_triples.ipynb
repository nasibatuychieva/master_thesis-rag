```python
def norm(s):
    if pd.isna(s):
        return ""
    return str(s).strip().lower()

# Gold normalisieren
gold_norm = gold.copy()
gold_norm["subject_n"]  = gold_norm["subject"].apply(norm)
gold_norm["relation_n"] = gold_norm["relation"].apply(norm)
gold_norm["object_n"]   = gold_norm["object"].apply(norm)

# Predicted normalisieren
pred_norm = pred.copy()
pred_norm["subject_n"]  = pred_norm["subject"].apply(norm)
pred_norm["relation_n"] = pred_norm["relation"].apply(norm)
pred_norm["object_n"]   = pred_norm["object"].apply(norm)
```

```python
def make_key(df, chunk_col):
    return (
        df[chunk_col].astype(str).str.strip().str.lower()
        + "||" + df["subject_n"]
        + "||" + df["relation_n"]
        + "||" + df["object_n"]
    )

gold_norm["triple_key"] = make_key(gold_norm, "chunk_id")
pred_norm["triple_key"] = make_key(pred_norm, "chunk_id")
```

```python
gold_keys = set(gold_norm["triple_key"].dropna().unique())
pred_norm = pred_norm.drop_duplicates(subset=["triple_key", "source"])
```

```python
def evaluate_source(gold_df, pred_df, source_name):
    """
    gold_df: normalisiertes Gold-DataFrame  (mit triple_key)
    pred_df: normalisiertes Pred-DataFrame  (mit triple_key)
    source_name: Name in Spalte 'source' (z.B. 'llama_index')
    """
    gold_keys = set(gold_df["triple_key"].dropna().unique())
    pred_source = pred_df[pred_df["source"] == source_name].copy()
    pred_keys = set(pred_source["triple_key"].dropna().unique())
    
    tp = len(gold_keys & pred_keys)
    fp = len(pred_keys - gold_keys)
    fn = len(gold_keys - pred_keys)
    
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1        = (2 * precision * recall / (precision + recall)) if (precision + recall) > 0 else 0.0
    
    return {
        "source": source_name,
        "tp": tp,
        "fp": fp,
        "fn": fn,
        "precision": precision,
        "recall": recall,
        "f1": f1,
    }

sources = pred_norm["source"].unique()
results = [evaluate_source(gold_norm, pred_norm, s) for s in sources]
eval_df = pd.DataFrame(results)
eval_df
```

```python
def evaluate_source_per_chunk(gold_df, pred_df, source_name):
    gold_keys_chunk = gold_df.groupby("chunk_id")["triple_key"].apply(set)
    pred_source = pred_df[pred_df["source"] == source_name]
    pred_keys_chunk = pred_source.groupby("chunk_id")["triple_key"].apply(set)

    rows = []
    all_chunks = sorted(set(gold_keys_chunk.index) | set(pred_keys_chunk.index))
    
    for ch in all_chunks:
        gk = gold_keys_chunk.get(ch, set())
        pk = pred_keys_chunk.get(ch, set())
        tp = len(gk & pk)
        fp = len(pk - gk)
        fn = len(gk - pk)
        prec = tp / (tp + fp) if (tp + fp) > 0 else 0.0
        rec  = tp / (tp + fn) if (tp + fn) > 0 else 0.0
        f1   = (2 * prec * rec / (prec + rec)) if (prec + rec) > 0 else 0.0
        rows.append({
            "source": source_name,
            "chunk_id": ch,
            "tp": tp, "fp": fp, "fn": fn,
            "precision": prec,
            "recall": rec,
            "f1": f1,
        })
    return pd.DataFrame(rows)

per_chunk_simplekg = evaluate_source_per_chunk(gold_norm, pred_norm, "simplekg")
per_chunk_simplekg.head()
```
