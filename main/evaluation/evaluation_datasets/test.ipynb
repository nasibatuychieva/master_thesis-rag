```python
import pandas as pd
from pathlib import Path
import os

PROJECT_ROOT = Path(os.getenv("PROJECT_ROOT", Path.cwd())).expanduser().resolve()

# Input JSONL
INPUT_PATH = PROJECT_ROOT / "golden_answers_dataset.jsonl"

# Output JSONL
OUTPUT_PATH = PROJECT_ROOT / "golden_answers_dataset_filtered_rest.jsonl"

# Load
df = pd.read_json(INPUT_PATH, lines=True)

# Filter
keep_types = {"factual", "disambiguation", "reasoning"}
df_filtered = df[df["query_type"].isin(keep_types)].copy()


cols = ["id", "query_type", "question", "gold_answer"]
df_filtered = df_filtered[cols]

# Save as JSONL
df_filtered.to_json(OUTPUT_PATH, orient="records", lines=True, force_ascii=False)

print(f"Saved {len(df_filtered)} rows to: {OUTPUT_PATH}")

# Reload check
df_check = pd.read_json(OUTPUT_PATH, lines=True)
df_check.head()
```
