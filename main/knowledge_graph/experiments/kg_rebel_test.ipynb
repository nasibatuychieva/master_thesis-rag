```python
import spacy, re
nlp = spacy.load("en_core_web_sm")

def triples_spacy(text: str):
    doc = nlp(text)
    out = []
    for sent in doc.sents:
        subj = None; verb = None; obj = None
        for tok in sent:
            if tok.dep_ in ("nsubj", "nsubjpass"): subj = tok.subtree
            if tok.pos_ == "VERB" and tok.dep_ in ("ROOT","conj"): verb = tok
            if tok.dep_ in ("dobj","pobj","attr"): obj = tok.subtree
        s = " ".join(t.text for t in subj) if subj else ""
        p = re.sub(r"[^a-z0-9]+","_", (verb.lemma_ if verb else "").lower()).strip("_")
        o = " ".join(t.text for t in obj) if obj else ""
        if s and p and o:
            out.append([s.strip(), p, o.strip()])
    return out

triples=(triples_spacy(segment))
# Beispielausgabe (ungefähr):
# [['The Arduino Uno', 'be', 'based on the ATmega328P microcontroller'],
#  ['The Arduino Uno', 'connect', 'a computer']]
```

```python
from transformers import pipeline
import json

# ------------------------------------------------------------
# Globale Variable für Lazy Loading des REBEL-Modells
# ------------------------------------------------------------
triplet_extractor = None

def load_pipeline():
    """
    Lädt die Babelscape/REBEL Pipeline (nur einmal).
    """
    global triplet_extractor
    if triplet_extractor is None:
        print("📦 Modell-Pipeline wird geladen...")
        triplet_extractor = pipeline(
            'text2text-generation',
            model='Babelscape/rebel-large',
            tokenizer='Babelscape/rebel-large'
        )
        print(" Modell-Pipeline geladen.")

# ------------------------------------------------------------
# Richtiger Parser für REBEL (mit <rel>-Tag)
# ------------------------------------------------------------
def parse_triplets(generated: str):
    """
    Parser für Babelscape/REBEL (<triplet> <subj> ... <obj> ... <rel> ...).
    Robust gegen mehrere Tripel hintereinander.
    """
    toks = (
        generated.replace("<s>", "")
                 .replace("</s>", "")
                 .replace("<pad>", "")
                 .split()
    )

    triples = []
    subj, obj, rel = "", "", ""
    cur = None  # 'subj' | 'obj' | 'rel' | None

    def maybe_flush():
        nonlocal subj, obj, rel
        if subj.strip() and obj.strip() and rel.strip():
            triples.append({
                "head": subj.strip(),
                "type": rel.strip(),
                "tail": obj.strip()
            })
        subj = obj = rel = ""

    i = 0
    while i < len(toks):
        t = toks[i]
        if t == "<triplet>":
            # schließe ggf. das vorige Tripel
            maybe_flush()
            cur = None
        elif t == "<subj>":
            cur = "subj"
        elif t == "<obj>":
            cur = "obj"
        elif t == "<rel>":
            cur = "rel"
        else:
            if cur == "subj":
                subj += (" " if subj else "") + t
            elif cur == "obj":
                obj  += (" " if obj  else "") + t
            elif cur == "rel":
                rel  += (" " if rel  else "") + t
        i += 1

    maybe_flush()
    return triples

# ------------------------------------------------------------
# Robuste Extraktionsfunktion mit Fallback
# ------------------------------------------------------------
def extract_relations(text: str):
    """
    Ruft das REBEL-Modell auf, dekodiert den Output und gibt Triplets zurück.
    Funktioniert mit alten und neuen Transformers-Versionen.
    """
    load_pipeline()

    try:
        # Neuerer Transformers-Stil (liefert "generated_text")
        response = triplet_extractor(
            text,
            num_beams=6,
            max_new_tokens=256,
            no_repeat_ngram_size=3,
            length_penalty=1.05,
            early_stopping=True,
            return_text=True
        )
        raw = response[0]["generated_text"]

    except Exception:
        # Älterer Stil (liefert Token-IDs)
        response = triplet_extractor(
            text,
            num_beams=6,
            max_new_tokens=256,
            no_repeat_ngram_size=3,
            length_penalty=1.05,
            early_stopping=True,
            return_tensors=True,
            return_text=False
        )
        raw = triplet_extractor.tokenizer.batch_decode(
            [response[0]["generated_token_ids"]]
        )[0]

    return parse_triplets(raw)

# ------------------------------------------------------------
# Test
# ------------------------------------------------------------

text = segment
def make_natural(text):
    # Beispielhaft einfache Heuristiken
    text = text.replace("has feature", "includes the feature")
    text = text.replace("has power feature", "is equipped with the power feature")
    text = text.replace("is compatible with", "is compatible with the standard")
    return text

triples = extract_relations(make_natural(text))
print(json.dumps(triples, indent=2, ensure_ascii=False))
```

```python
# rebel_ok.py
from transformers import pipeline
import re

# 1) REBEL laden
rebel = pipeline(
    "text2text-generation",
    model="Babelscape/rebel-large",
    tokenizer="Babelscape/rebel-large",
    device_map="auto",
    torch_dtype="auto"
)

# 2) Parser: <triplet> <subj> SUBJECT <obj> OBJECT <rel> RELATION
def parse_rebel(generated: str):
    txt = generated.replace("<s>", "").replace("</s>", "").replace("<pad>", "").strip()
    triples = []
    for block in txt.split("<triplet>"):
        block = block.strip()
        if not block:
            continue
        m = re.search(r"<subj>(.*?)<obj>(.*?)<rel>(.*)", block, flags=re.DOTALL)
        if not m:
            continue
        subj = m.group(1).strip()
        obj  = m.group(2).strip()
        rel  = m.group(3).strip()
        if subj and rel and obj:
            triples.append({"head": subj, "type": rel, "tail": obj})
    # dedupe
    out, seen = [], set()
    for t in triples:
        k = (t["head"], t["type"], t["tail"])
        if k not in seen:
            out.append(t); seen.add(k)
    return out

# 3) Einzeiler: Text -> Tripel (ohne return_full_text!)
def extract_triples(text: str):
    gen = rebel(
        text,
        max_new_tokens=128,       # oder max_length=512
        num_beams=2,
        do_sample=False,
        clean_up_tokenization_spaces=False
    )[0]["generated_text"]
    return parse_rebel(gen)

# --- Test ---

txt = segment
print(extract_triples(txt))
```

```python
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

def extract_triplets(text):
    triplets = []
    relation, subject, relation, object_ = '', '', '', ''
    text = text.strip()
    current = 'x'
    for token in text.replace("<s>", "").replace("<pad>", "").replace("</s>", "").split():
        if token == "<triplet>":
            current = 't'
            if relation != '':
                triplets.append({'head': subject.strip(), 'type': relation.strip(),'tail': object_.strip()})
                relation = ''
            subject = ''
        elif token == "<subj>":
            current = 's'
            if relation != '':
                triplets.append({'head': subject.strip(), 'type': relation.strip(),'tail': object_.strip()})
            object_ = ''
        elif token == "<obj>":
            current = 'o'
            relation = ''
        else:
            if current == 't':
                subject += ' ' + token
            elif current == 's':
                object_ += ' ' + token
            elif current == 'o':
                relation += ' ' + token
    if subject != '' and relation != '' and object_ != '':
        triplets.append({'head': subject.strip(), 'type': relation.strip(),'tail': object_.strip()})
    return triplets

# Load model and tokenizer
tokenizer = AutoTokenizer.from_pretrained("Babelscape/rebel-large")
model = AutoModelForSeq2SeqLM.from_pretrained("Babelscape/rebel-large")
gen_kwargs = {
    "max_length": 256,
    "length_penalty": 0,
    "num_beams": 3,
    "num_return_sequences": 3,
}

# Text to extract triplets from
text = segment

# Tokenizer text
model_inputs = tokenizer(text, max_length=256, padding=True, truncation=True, return_tensors = 'pt')

# Generate
generated_tokens = model.generate(
    model_inputs["input_ids"].to(model.device),
    attention_mask=model_inputs["attention_mask"].to(model.device),
    **gen_kwargs,
)

# Extract text
decoded_preds = tokenizer.batch_decode(generated_tokens, skip_special_tokens=False)

# Extract triplets
for idx, sentence in enumerate(decoded_preds):
    print(f'Prediction triplets sentence {idx}')
    print(extract_triplets(sentence))
```
